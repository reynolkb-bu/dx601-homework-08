### Problem 16

Set `p16` to the first principal component of the abalone data set.
Give it in the form of a one row data frame like in the Palmer Penguins example.

Hint: Don't forget the example code from the videos is linked in the repositories linked above.

```python
# YOUR CHANGES HERE
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

abalone_numeric = abalone.select_dtypes(include=["number"])
X = abalone_numeric.drop("Rings", axis=1)
y = abalone_numeric["Rings"]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=1)
X_pca = pca.fit_transform(X_scaled)

loadings = pd.DataFrame( pca.components_.T, columns=['PC1'], index=X.columns )
print("Feature contributions to each PC:")
print(loadings.round(3))

p16 = X_pca[:, 0]
p16
```
